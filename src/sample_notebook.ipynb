# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
run_mode = dbutils.widgets.get("run_mode")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
# Inspect taxi samples with a run-mode aware limit
limit = 10 if run_mode == "sample" else None
df = spark.read.table("samples.nyctaxi.trips")
if limit:
    df = df.limit(limit)
df.select("pickup_datetime", "dropoff_datetime", "trip_distance").show(limit or 10)
```
